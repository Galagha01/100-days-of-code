```python
import pandas as pd
import numpy as np

# Customer information DataFrame
customers = pd.DataFrame({
    'CustomerID': [101, 102, 103, 104, 105, 106],
    'Name': ['Alice Johnson', 'Bob Smith', 'Carol White', 'David Brown', 'Eve Davis', 'Frank Miller'],
    'City': ['NYC', 'LA', 'Chicago', 'Houston', 'Phoenix', 'NYC']
})

# Order information DataFrame
orders = pd.DataFrame({
    'OrderID': [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008],
    'CustomerID': [101, 102, 101, 105, 103, 107, 102, 104],
    'OrderDate': ['2024-01-15', '2024-01-16', '2024-01-18', '2024-01-20',
                  '2024-01-22', '2024-01-23', '2024-01-25', '2024-01-26'], # Fixed: Added missing closing single quote
    'TotalAmount': [250.00, 180.00, 320.00, 150.00, 420.00, 90.00, 275.00, 380.00]
})

# Product details DataFrame
products = pd.DataFrame({
    'ProductID': ['P1', 'P2', 'P3', 'P4', 'P5'],
    'ProductName': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Headphones'],
    'Price': [800, 25, 75, 300, 150]
})

# Order items (which products were in each order) DataFrame
order_items = pd.DataFrame({
    'OrderID': [1001, 1001, 1002, 1003, 1003, 1004, 1005, 1006, 1007, 1008],
    'ProductID': ['P1', 'P2', 'P2', 'P1', 'P4', 'P5', 'P3', 'P2', 'P4', 'P1'],
    'Quantity': [1, 2, 3, 1, 1, 1, 2, 1, 1, 1]
})

print("Customers DataFrame:")
display(customers)
print("\nOrders DataFrame:")
display(orders)
print("\nProducts DataFrame:")
display(products)
print("\nOrder Items DataFrame:")
display(order_items)
```

```python
# Perform an inner merge between 'customers' and 'orders' DataFrames on 'CustomerID'.
# An inner merge combines two DataFrames based on a common column(s)
# and returns only the rows that have matching values in both DataFrames.
# This merge specifically shows customers who have placed orders.
innerMerge = pd.merge(customers, orders, on='CustomerID', how='inner')

print("Inner Merge Result (Customers with matching orders):")
display(innerMerge)
```

```python
# Perform a left merge, keeping all rows from the 'orders' DataFrame.
# This merge combines 'orders' with 'customers' on 'CustomerID'.
# For Order 1006, there is no matching customer (CustomerID 107 is not in 'customers'),
# so customer details (Name, City) will be NaN (Not a Number).
leftMerge = pd.merge(orders, customers, on='CustomerID', how='left')

print("Left Merge Result (All orders with customer details if available):")
display(leftMerge)
```

```python
# Perform an outer merge, keeping all rows from both 'orders' and 'customers' DataFrames.
# The 'indicator=True' argument adds a column '_merge' indicating the source of each row:
# - 'both' means the row exists in both DataFrames.
# - 'left_only' means the row exists only in the left DataFrame ('orders').
# - 'right_only' means the row exists only in the right DataFrame ('customers').
outerMerge = pd.merge(orders, customers, on='CustomerID', how='outer', indicator=True)

print("Outer Merge Result (All orders and all customers):")
display(outerMerge)

# Observations from the outer merge:
# - Customer Frank Miller (CustomerID 106) has no orders, so their order details are NaN ('right_only').
# - Order 1006 (CustomerID 107) has no matching customer details, so customer info is NaN ('left_only').
```

```python
# --- Task 4: Multiple Merges --- (2 points)
# Combine multiple DataFrames to create a comprehensive report:

# 1. Start by merging order_items with products to get product names and prices.
# An inner merge is used to ensure only valid ProductIDs are included.
pidMerge = pd.merge(order_items, products, on='ProductID', how='inner')

# 2. Then merge the result (pidMerge) with orders to get order dates and totals.
# An inner merge is used here to link order details to the product items.
oidMerge = pd.merge(pidMerge, orders, on='OrderID', how='inner')

# 3. Display the first 10 rows of the final DataFrame.
# The result should show: OrderID, ProductID, ProductName, Price, Quantity, OrderDate, TotalAmount
print("Combined Order Items, Products, and Orders (first 10 rows of comprehensive report):")
display(oidMerge.head(10))
```

```python
# --- Task: Summarize total spending per customer --- (Original cell before comprehensive fix)

# Merge 'customers' with 'orders' on 'CustomerID' to link customer information with their orders.
# An inner merge is used to include only customers who have placed orders.
cidrMerge = pd.merge(customers, orders, on='CustomerID', how='inner')

# Group the merged DataFrame by 'Name' and 'City' and calculate the total amount spent by each customer.
cidGroupBy = cidrMerge.groupby(['Name', 'City'])['TotalAmount'].sum()

# Sort the results by 'TotalAmount' in descending order to identify top spenders.
# 'ascending=False' ensures descending order.
cidSort = cidGroupBy.sort_values(ascending=False, na_position='last')

print("Total spending per customer (sorted by total amount - original output style):")
display(cidGroupBy)
```

```python
# --- Task: Summarize total spending per customer (Corrected Sorting) ---

# Group the merged DataFrame by 'Name' and 'City' and calculate the total amount spent by each customer.
# This re-uses 'cidrMerge' from the previous step.
cidGroupBy = cidrMerge.groupby(['Name', 'City'])['TotalAmount'].sum()

# Sort the results by 'TotalAmount' in descending order to identify top spenders.
# This correction uses 'ascending=False' for the Series.sort_values method.
cidSort = cidGroupBy.sort_values(ascending=False, na_position='last')

print("Total spending per customer (sorted descending):")
display(cidSort)

# Answer to the question:
# This output clearly shows the top spenders. Alice Johnson from NYC is the top spender with $570.0.
```
